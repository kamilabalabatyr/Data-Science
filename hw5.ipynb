## A2 ★★ Оклад под защитой · 🖥 слайд 7 · 📓 шаг 7 (`Product`, `Account`)
`salary` — свойство с проверкой: ноль или минус → `ValueError`. Проверка должна срабатывать и при создании, и при изменении (`e.salary = -5`).

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"{self.name}: оклад {self.salary}"

# ваш код: превратите salary в свойство с setter
    @property
    def salary(self):
      return self._salary

    @salary.setter
    def salary(self, s):
      if s <=0:
        raise ValueError("Цена должна быть больше нуля")
      self._salary = s
```

```python
# ✅ Проверка A2 — просто запустите ячейку
try:
    e = Employee("Аня", 3000)
    e.salary = 3500
    assert e.salary == 3500
    for bad in (0, -1):
        try:
            Employee("X", bad)
            raise AssertionError(f"Employee('X', {bad}) должен дать ValueError")
        except ValueError:
            pass
    try:
        e.salary = -5
        raise AssertionError("e.salary = -5 должен дать ValueError")
    except ValueError:
        pass
    assert e.salary == 3500, "после неудачной попытки оклад не должен измениться"
    print("✅ A2: всё верно!")
except AssertionError as e:
    print("❌ A2: пока не так —", e)
except Exception as e:
    print(f"❌ A2: ошибка {type(e).__name__}: {e}")
```

🗣 **Почему так?** Почему в `__init__` пишем `self.salary = salary` (без подчёркивания), а не `self._salary = salary`? Что будет с `Employee("X", -1)` во втором случае? (Это та же ошибка, что в задаче 7.2 тетради.)

**Ваш ответ:** Потому что self.salary = salary идёт через setter, и проверка value <= 0 срабатывает уже при создании объекта. Если написать self._salary = salary, значение записывается напрямую, мимо setter, и проверки нет. Тогда Employee("X", -1) создастся без ошибки с окладом -1, и объект с самого начала будет некорректным. Защита сработает только при следующем изменении (e.salary = -5), но «плохой» сотрудник уже существует.

## A4 ★★ `Manager` · 🖥 слайды 8–9 · 📓 шаг 8
`Manager("Айгерим", 5000, bonus=1000)`: выплата = оклад + премия. Наследуйтесь от `Employee`, не забудьте `super().__init__`.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary              # идёт через setter — с проверкой

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value <= 0:
            raise ValueError("Оклад должен быть больше нуля")
        self._salary = value

    def __str__(self):
        return f"{self.name}: оклад {self.salary}"

    def monthly_pay(self):
        return self.salary


class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def monthly_pay(self):
        return self.salary + self.bonus
```

```python
# ✅ Проверка A4 — просто запустите ячейку
try:
    m = Manager("Айгерим", 5000, bonus=1000)
    assert m.monthly_pay() == 6000, f"5000 + 1000 = 6000, получилось {m.monthly_pay()}"
    assert isinstance(m, Employee), "Manager должен наследовать Employee"
    assert str(m) == "Айгерим: оклад 5000", "__str__ достаётся от Employee"
    try:
        Manager("X", -1, 100)
        raise AssertionError("проверка оклада должна работать и у менеджера")
    except ValueError:
        pass
    print("✅ A4: всё верно!")
except AssertionError as e:
    print("❌ A4: пока не так —", e)
except Exception as e:
    print(f"❌ A4: ошибка {type(e).__name__}: {e}")
```

🗣 **Почему так?** Проверку оклада в `Manager` вы не писали, но `Manager("X", -1, 100)` всё равно падает с `ValueError`. Откуда она взялась? Что случится, если в `Manager.__init__` забыть `super().__init__(...)`?

**Ваш ответ:** Потому что ему эта валидация присвоится через родитель класса Employee. super().__init_(...) вот эта строка как раз таки вызывает self.salary = salary, via setter, и получается они будут проходит проверку. Если забыть написать super().__init__() Employee.__init__ не выполнится: проверки не будет, Manager("X", -1, 100) создастся без ошибки, но у объекта не будет name и _salary, и позже monthly_pay() или str(m) упадут с AttributeError.

## A5 ★★ `Developer` · 🖥 слайд 9 · 📓 шаг 8
`Developer("Тимур", 2500, "Python", overtime_hours=20)`: выплата = оклад + `overtime_hours * 10`; `overtime_hours` по умолчанию 0.
`__str__` **дополняет** родительский через `super()`: `"Тимур: оклад 2500, Python"`.

```python
class Developer(Employee):
    def __init__(self, name, salary, language, overtime_hours=0):
        super().__init__(name, salary)
        self.language = language
        self.overtime_hours = overtime_hours

    @property
    def overtime_hours(self):
      return self._overtime_hours

    @overtime_hours.setter
    def overtime_hours(self, h):
      if h<0:
        raise ValueError("Hour value must be 0 or higher")
      self._overtime_hours = h

    def monthly_pay(self):
        return self.salary + (self.overtime_hours*10)

    def __str__(self):
        base = super().__str__()
        return base + f", {self.language}"
```

```python
# ✅ Проверка A5 — просто запустите ячейку
try:
    d = Developer("Тимур", 2500, "Python", overtime_hours=20)
    assert d.monthly_pay() == 2700, f"2500 + 20 * 10 = 2700, получилось {d.monthly_pay()}"
    assert Developer("Дана", 2700, "Go").monthly_pay() == 2700, "без переработок — просто оклад"
    assert str(d) == "Тимур: оклад 2500, Python", f"получилось {str(d)!r}"
    print("✅ A5: всё верно!")
except AssertionError as e:
    print("❌ A5: пока не так —", e)
except Exception as e:
    print(f"❌ A5: ошибка {type(e).__name__}: {e}")
```

## B3 ★★ Погода: 30 дней × 3 города · 🖥 слайды 25–26 · 📓 шаги N9, N10
Строки — дни, столбцы — города.
1. `city_mean` — средняя температура по каждому городу.
2. `city_max` — максимум по каждому городу, `hot_day` — **номер дня** этого максимума (найдите в документации `np.argmax`, параметр `axis`).
3. `hot_days` — сколько дней жарче 25 °C в каждом городе.
4. `z` — нормализация по столбцам: `(temps − среднее столбца) / std столбца`.

```python
import numpy as np
temps = np.random.default_rng(0).normal(20, 5, size=(30, 3)).round(1)
print(temps.shape)
print(temps[:5])        # первые 5 дней
```

```python
city_mean = temps.mean(axis=0)
city_max = temps.max(axis=0)
hot_day = temps.argmax(axis=0)
hot_days = (temps > 25).sum(axis=0)
z = (temps - temps.mean(axis=0)) / temps.std(axis=0)
```

```python
# ✅ Проверка B3 — просто запустите ячейку
try:
    assert city_mean.shape == (3,), "по одному среднему на город — форма (3,). Какая axis «схлопывает» дни?"
    assert np.allclose(city_mean.round(2), [19.65, 20.18, 21.82]), f"city_mean: {city_mean.round(2)}"
    assert np.allclose(city_max, [29.0, 30.0, 29.8]), f"city_max: {city_max}"
    assert list(hot_day) == [16, 26, 15], f"hot_day: {hot_day}"
    assert list(hot_days) == [7, 5, 5], f"hot_days: {hot_days}"
    assert z.shape == (30, 3) and np.allclose(z.mean(axis=0), 0) and np.allclose(z.std(axis=0), 1), "у каждого столбца z: среднее 0, std 1"
    print("✅ B3: всё верно!")
except AssertionError as e:
    print("❌ B3: пока не так —", e)
except Exception as e:
    print(f"❌ B3: ошибка {type(e).__name__}: {e}")
```

Каждый способ замерьте по шаблону ниже. Все три должны дать **одно и то же число**.

```python
import time

start = time.perf_counter()
total_loop = 0
for i in range(1, 1_000_001):
    total_loop += i * i
t_loop = time.perf_counter() - start

start = time.perf_counter()
total_gen = sum(i * i for i in range(1, 1_000_001))
t_gen = time.perf_counter() - start

start = time.perf_counter()
total_np = (np.arange(1, 1_000_001, dtype=np.int64) ** 2).sum()
t_np = time.perf_counter() - start

print(f"цикл: {t_loop * 1000:.1f} мс, генератор: {t_gen * 1000:.1f} мс, numpy: {t_np * 1000:.1f} мс")
```

## C1 Класс `Sensor`
`Sensor("Алматы", values)` хранит имя и **массив Numpy** (`np.array(values, dtype=float)`). Методы:
- `mean()` — среднее;
- `above(threshold)` — **булева маска** значений выше порога;
- `count_above(threshold)` — сколько таких значений (используйте `above`!).
📓 шаг N12 — класс `Gradebook` устроен так же.

```python
class Sensor:
    # датчик: имя города и массив измерений
    def __init__(self, name, values):
        self.name = name
        self.values = np.array(values, dtype=float)

    def mean(self):
        return self.values.mean()

    def above(self, threshold):
        return self.values > threshold        # булева маска

    def count_above(self, threshold):
        return self.above(threshold).sum()    # True считается как 1
```

```python
# ✅ Проверка C1 — просто запустите ячейку
try:
    s = Sensor("Алматы", temps[:, 0])
    assert np.isclose(s.mean(), temps[:, 0].mean())
    assert s.above(25).dtype == bool and s.above(25).shape == (30,), "above возвращает маску той же длины"
    assert s.count_above(25) == 7
    print("✅ C1: всё верно!")
except AssertionError as e:
    print("❌ C1: пока не так —", e)
except Exception as e:
    print(f"❌ C1: ошибка {type(e).__name__}: {e}")
```

🖥 слайд 11 · 📓 шаг 10.

```python
from abc import ABC, abstractmethod
import json

class BaseEmployee(ABC):
    # абстрактный сотрудник: создать напрямую нельзя
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @abstractmethod
    def monthly_pay(self):
        pass

class JsonMixin:
    # миксин: умеет выгружать атрибуты объекта в JSON
    def to_json(self):
        data = {k.lstrip("_"): v for k, v in self.__dict__.items()}
        return json.dumps(data, ensure_ascii=False)

class JsonManager(JsonMixin, Manager):
    pass
```
