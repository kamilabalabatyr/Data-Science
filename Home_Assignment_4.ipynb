**Задача 6. Продажи за неделю**

```python
days = ["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"]
sales = [280, 310, 295, 420, 380, 150, 330]

plan = 300

total = 0
best_day = days[0]
best_sale = sales[0]
plan_done = 0
plan_failed = []

for i in range(len(sales)):
    total += sales[i]                 # 1. накапливаем сумму

    if sales[i] > best_sale:          # 2. ищем лучший день вручную
        best_sale = sales[i]
        best_day = days[i]

    if sales[i] >= plan:              # 3. план выполнен
        plan_done += 1
    else:                             # 4. план не выполнен
        plan_failed.append(days[i])

average = total / len(sales)

print(f"Сумма за неделю: {total:,} тыс. тг")
print(f"Среднее за день: {average:.2f} тыс. тг")
print(f"Лучший день: {best_day} ({best_sale} тыс. тг)")
print(f"План выполнен: {plan_done} дн.")
print(f"План не выполнен: {', '.join(plan_failed)}")

```

**Мини-проект. Выгрузка из урока 3 — теперь циклом**

```python
names = ["Дана", "Ерлан", "Жанна", "Бекзат", "Айгерим"]
ages = ["19", "", "21", None, "20"]
payments = ["1 234,50 тг", "999 тг", "N/A", "15 000,00тг", "2 500 тг"]
scholarships = ["нет", "да", "0", "1", "Да "]
cities = [" астана", "АСТАНА ", "Алматы", "алматы", "Шымкент"]

missings = ["" , None , "N/A"]
clean_ages, clean_payments, clean_scholarships, clean_cities = [], [], [], []

#1. В одном цикле пройдите по всем студентам и очистите возраст, оплату, стипендию и город. Пропуски ( "" , None , "N/A" ) обработайте через if .
for name, age, payment, scholarship, city in zip(names, ages, payments, scholarships, cities):
  if age not in missings:
    clean_ages.append(int(age))
  else:
    clean_ages.append(None)

  if payment not in missings:
    p = payment.replace("тг", "").replace(" ", "").replace(",", ".")
    clean_payments.append(float(p))
  else:
    clean_payments.append(None)

  if scholarship not in missings:
    clean_scholarships.append(scholarship.strip().lower()in ["да", "1"])
  else:
    clean_scholarships.append(None)

  if city not in missings:
    c = city.replace(" ", "").title()
    clean_cities.append(c)
  else:
    clean_cities.append(None)

#2. Выведите таблицу с выровненными столбцами. Пропуски покажите знаком «—».
print(f"{'NAME':^10} | {'AGE':^10} | {'PAYMENT':^10} | {'SCHOLARSHIP':^15} | {'CITY':^10}")
print("-"*69)
for i in range(len(names)):
  age_t = str(clean_ages[i]) if clean_ages[i] is not None else "-"
  pay_t = f"{clean_payments[i]:,.2f}" if clean_payments[i] is not None else "-"
  if clean_scholarships[i] is None:
      sch_t = "—"
  else:
      sch_t = "да" if clean_scholarships[i] else "нет"
  city_t = str(clean_cities[i]) if clean_cities[i] is not None else "-"

  print(f"{names[i]:<10} | {age_t:^10} | {pay_t:>10} | {sch_t:^15} | {city_t:^10} |")


#3. Выведите итоги: средний возраст, сумму оплат, число студентов со стипендией, число уникальных городов.
age_sum = 0
age_count = 0       # сколько возрастов известно
pay_sum = 0
sch_count = 0
unique_cities = []
for i in range(len(names)):
  if clean_ages[i] is not None:
    age_sum += clean_ages[i]
    age_count += 1

  if clean_payments[i] is not None:
    pay_sum += clean_payments[i]

  if clean_scholarships[i] == True:
    sch_count+=1

  if clean_cities[i] is not None and clean_cities[i] not in unique_cities:
        unique_cities.append(clean_cities[i])

avg_age = age_sum/age_count

print("-"*69)
print(f"{'TOTAL:':<10} | {avg_age:^10.2f} | {pay_sum:>10,.2f} | {sch_count:^15} | {len(unique_cities):^10} |")
```

```python
#Задача 12. Матрица продаж
shops = ["Магазин 1", "Магазин 2", "Магазин 3"]
sales = [[120, 340, 210, 180],
         [95, 400, 180, 220],
         [300, 150, 260, 310]]

# Сумма по магазинам (идём по строкам)
shop_totals = []
for i in range(len(sales)):
    total = 0
    for j in range(len(sales[i])):
        total += sales[i][j]
    shop_totals.append(total)
    print(f"{shops[i]}: {total} тыс. тг")

print()

# Сумма по дням (идём по столбцам: внешний цикл по дням, внутренний по магазинам)
for j in range(len(sales[0])):
    total = 0
    for i in range(len(sales)):
        total += sales[i][j]
    print(f"День {j + 1}: {total} тыс. тг")

# Лучший магазин
best = 0
for i in range(1, len(shop_totals)):
    if shop_totals[i] > shop_totals[best]:
        best = i
print(f"\nЛучший магазин: {shops[best]} ({shop_totals[best]} тыс. тг)")
```

```python
#r3
passwords = ["qwerty", "Python2026", "password123", "SECRET99"]

for pwd in passwords:
    has_digit = False
    has_upper = False
    for ch in pwd:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True

    reasons = []
    if len(pwd) < 8:
        reasons.append("меньше 8 символов")
    if not has_digit:
        reasons.append("нет цифры")
    if not has_upper:
        reasons.append("нет заглавной буквы")

    if reasons:
        print(f"{pwd}: ненадёжный — {', '.join(reasons)}")
    else:
        print(f"{pwd}: надёжный")
```

```python
#r4
n = 27
steps = 0
max_value = n

while n != 1:
    if n % 2 == 0:
        n = n // 2           # // — целочисленное деление, чтобы не получить 41.0
    else:
        n = n * 3 + 1
    steps += 1
    if n > max_value:
        max_value = n

print(f"Шагов до 1: {steps}")
print(f"Наибольшее число: {max_value}")
```
