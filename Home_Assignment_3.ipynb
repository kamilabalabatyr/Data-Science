Вторая часть: одним выражением проверьте, является ли год високосным: он делится на 4, но не делится на 100 — либо делится на 400. Проверьте годы 2024, 2026, 1900, 2000.

```python
year = [2024, 2026, 1900, 2000]
for y in year:
  vis = "Yes, it is leap year " if y % 4 == 0 and (y % 100 != 0 or y % 400 == 0) else "It is not a leap year"
  print(y, "->", vis)
```

**Мини-проект. Очистка выгрузки**

```python
rows = [
{"name": "Дана", "age": "19", "payment": "1 234,50 тг", "scholarship": "нет", "city": " астана"},
{"name": "Ерлан", "age": "", "payment": "999 тг", "scholarship": "да", "city": "АСТАНА "},
{"name": "Жанна", "age": "21", "payment": "N/A", "scholarship": "0", "city": "Алматы"},
{"name": "Бекзат", "age": None, "payment": "15 000,00тг", "scholarship": "1", "city": "алматы"},
]

#1. Возраст преобразуйте в int , пропуски — в None .
for row in rows:
    age = row["age"]
    row["age"] = int(age) if age else None

missing_payments = 0
#2. Оплату преобразуйте в float , «N/A» — в None .
missing = {"", "N/A", "NA", "-", "NULL"}
for row in rows:
    payment = row["payment"]
    if payment is None or payment.strip().upper() in missing:
        row["payment"] = None
        missing_payments += 1
    else:
        row["payment"] = float(payment.replace("тг", "").replace(" ", "").replace(",", "."))

with_scholarship = 0
#3. Стипендию преобразуйте в bool : «да» и «1» означают True.
for row in rows:
    row["scholarship"] = bool(row["scholarship"].strip().lower() == "да" or row["scholarship"].strip().lower() == "1")
    if row["scholarship"] == True:
      with_scholarship +=1

#4. Приведите названия городов к единому виду.
for row in rows:
      row["city"] = row["city"].strip().title()

#5. Выведите: средний возраст, сумму оплат и число пропусков, количество студентов со стипендией, число уникальных городов.
ages = [r["age"] for r in rows if r["age"] is not None]
avg_age = sum(ages) / len(ages)

payments = [r["payment"] for r in rows if r["payment"] is not None]
total_payment = sum(payments)

unique_cities = len({r["city"] for r in rows})

print(f"Средний возраст:        {avg_age}")
print(f"Сумма оплат:            {total_payment:,.2f} тг")
print(f"Пропусков в оплате:     {missing_payments}")
print(f"Со стипендией:          {with_scholarship}")
print(f"Уникальных городов:     {unique_cities}")
#6. Выведите очищенную таблицу с выравниванием столбцов; пропуски покажите знаком «—».
print(f"{'Имя':<8} | {'Возраст':>7} | {'Оплата':>10} | {'Стипендия':>10} | {'Город':<8} |")
print("-" * 57)

for r in rows:
    age = r["age"] if r["age"] is not None else "—"
    payment = f"{r['payment']:.2f}" if r["payment"] is not None else "—"
    scholarship = "да" if r["scholarship"] else "нет"

    print(f"{r['name']:<8} | {age:>7} | {payment:>10} | {scholarship:^10} | {r['city']:^8} |")
    print("-" * 57)
```

**R1. Нормализация телефонa**

```python
phones = ["+7 (701) 123-45-67", "8 701 123 45 67", "87011234567"]

formatted_phones = []
for p in phones:
    digits = "".join(ch for ch in p if ch.isdigit())
    if digits.startswith("8"):
        digits = "7" + digits[1:]
    formatted_phones.append("+" + digits)

print(formatted_phones)
```
